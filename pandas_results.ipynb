```python
import pandas as pd
import matplotlib.pyplot as plt

dqn21 = pd.read_csv("./DQNResults/2 Layer/1/dqnlearn-stats-1layer.csv")
dqn22 = pd.read_csv("./DQNResults/2 Layer/2/dqnlearn-stats.csv")
dqn23 = pd.read_csv("./DQNResults/2 Layer/3/dqnlearn-stats.csv")
dqn24 = pd.read_csv("./DQNResults/2 Layer/4/dqnlearn-stats.csv")
dqn25 = pd.read_csv("./DQNResults/2 Layer/5/dqnlearn-stats.csv")
ave_dqn2 = dqn21[["Episode"]].copy()
ave_dqn2.loc[:,"Epsilon"] = dqn21["Epsilon"]
ave_dqn2.loc[:,"Score 1"] = dqn21["Score"]
ave_dqn2.loc[:,"Score 2"] = dqn22["Score"]
ave_dqn2.loc[:,"Score 3"] = dqn23["Score"]
ave_dqn2.loc[:,"Score 4"] = dqn24["Score"]
ave_dqn2.loc[:,"Score 5"] = dqn25["Score"]
ave_dqn2.loc[:,"Steps 1"] = dqn21["Total Steps"]
ave_dqn2.loc[:,"Steps 2"] = dqn22["Total Steps"]
ave_dqn2.loc[:,"Steps 3"] = dqn23["Total Steps"]
ave_dqn2.loc[:,"Steps 4"] = dqn24["Total Steps"]
ave_dqn2.loc[:,"Steps 5"] = dqn25["Total Steps"]
ave_dqn2.loc[:,"Death 1"] = (dqn21["Death"] == " Status.SNAKE").astype(int)
ave_dqn2.loc[:,"Death 2"] = (dqn22["Death"] == " Status.SNAKE").astype(int)
ave_dqn2.loc[:,"Death 3"] = (dqn23["Death"] == " Status.SNAKE").astype(int)
ave_dqn2.loc[:,"Death 4"] = (dqn24["Death"] == " Status.SNAKE").astype(int)
ave_dqn2.loc[:,"Death 5"] = (dqn25["Death"] == " Status.SNAKE").astype(int)
ave_dqn2["Average Score"] = (ave_dqn2["Score 1"] + ave_dqn2["Score 2"] + ave_dqn2["Score 3"] + ave_dqn2["Score 4"] + ave_dqn2["Score 5"]) / 5
ave_dqn2["MA Score"] = ave_dqn2["Average Score"].rolling(window=50).mean()
ave_dqn2["Average Steps"] = (ave_dqn2["Steps 1"] + ave_dqn2["Steps 2"] + ave_dqn2["Steps 3"] + ave_dqn2["Steps 4"] + ave_dqn2["Steps 5"]) / 5
ave_dqn2["MA Steps"] = ave_dqn2["Average Steps"].rolling(window=50).mean()
ave_dqn2["Average Death"] = (ave_dqn2["Death 1"] + ave_dqn2["Death 2"] + ave_dqn2["Death 3"] + ave_dqn2["Death 4"] + ave_dqn2["Death 5"]) / 5
ave_dqn2["MA Death"] = ave_dqn2["Average Death"].rolling(window=50).mean()
```

```python
plt.plot(ave_dqn2['Episode'], ave_dqn2['Average Score'], alpha=0.3, label='Mean Score', color='blue')
plt.plot(ave_dqn2['Episode'], ave_dqn2['MA Score'], alpha=1.0, label='50-Ep Moving Avg', color='black', linewidth=1)
plt.title('DQN 2-Layer Score Performance')
plt.xlabel('Episode')
plt.ylabel('Score')
plt.legend()
plt.grid(True, alpha=0.3)
```

```python
plt.plot(ave_dqn2['Episode'], ave_dqn2['Average Steps'], alpha=0.3, label='Mean Steps', color='blue')
plt.plot(ave_dqn2['Episode'], ave_dqn2['MA Steps'], alpha=1.0, label='50-Ep Moving Avg', color='black', linewidth=1)
plt.title('DQN 2-Layer Steps Performance')
plt.xlabel('Episode')
plt.ylabel('Total Steps')
plt.legend()
plt.grid(True, alpha=0.3)
```

```python
plt.plot(ave_dqn2['Episode'], ave_dqn2['Average Death'], alpha=0.3, label='Snake Body Death Ratio', color='blue')
plt.plot(ave_dqn2['Episode'], ave_dqn2['MA Death'], alpha=1.0, label='50-Ep Moving Avg', color='black', linewidth=1)
plt.title('DQN 2-Layer Snake Body to Wall Death Performance')
plt.xlabel('Episode')
plt.ylabel('Body to Wall Death Ratio')
plt.legend()
plt.grid(True, alpha=0.3)
```
